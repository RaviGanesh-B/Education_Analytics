```python
import pandas as pd
import requests

adult_literacy_df = pd.read_csv("https://ourworldindata.org/grapher/literacy-rate-adults.csv?v=1&csvType=full&useColumnShortNames=true",storage_options = {"user-agent":"Our world In Data data fetch/1.0 "})

adult_literacy_rate_df = pd.DataFrame(adult_literacy_df)

adult_literacy_rate_df
```

```python
import pandas as pd 
import requests

young_df = pd.read_csv("https://ourworldindata.org/grapher/literacy-rate-of-young-men-and-women.csv?v=1&csvType=full&useColumnShortNames=true",storage_options = {"user-agent":"Our World In Data fetch/1.0"})

young_literacy_rate_df = pd.DataFrame(young_df)

young_literacy_rate_df
```

```python
import pandas as pd
import requests
from io import StringIO

illiterate_df = pd.read_csv("https://ourworldindata.org/grapher/literate-and-illiterate-world-population.csv?v=1&csvType=full&useColumnShortNames=true",storage_options = {"user-agent":"Our World In Data data fetch/1.0"})

illiterate_population = pd.DataFrame(illiterate_df)

illiterate_population
```

```python
import pandas as pd
import requests
from io import StringIO

gdp_df = pd.read_csv("https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true",storage_options = {"user-agent":"Our World In Data data fetch/1.0"})

gdp_percapita = pd.DataFrame(gdp_df)

gdp_percapita
```

```python
import pandas as pd
import requests
from io import StringIO

average_scholling_df = pd.read_csv("https://ourworldindata.org/grapher/literacy-rates-vs-average-years-of-schooling.csv?v=1&csvType=full&useColumnShortNames=true",storage_options = {"user-agent":"Our World In Data data fetch/1.0"})

average_years_of_scholling = pd.DataFrame(average_scholling_df)


average_years_of_scholling
```

```python
#step one done 1️⃣ Dataset Collection 📥 done
```

```python
import pandas as pd
import requests

adult_young_df = pd.merge(adult_literacy_rate_df,young_literacy_rate_df, on = ["entity","year","code"])

adult_young_df = adult_young_df.dropna(subset=['code'])

adult_young_df['entity'] = adult_young_df['entity'].str.strip()

adult_young_df["owid_region"] = adult_young_df["owid_region"].fillna("unknown")

adult_young_df = adult_young_df.rename(columns = {
    "adult_literacy_rate__population_15plus_years__both_sexes__pct__lr_ag15t99":"adult_literacy_rate",
    "youth_literacy_rate__population_15_24_years__male__pct__lr_ag15t24_m":"youth_literacy_rate_male",
    "youth_literacy_rate__population_15_24_years__female__pct__lr_ag15t24_f":"youth_literacy_rate_female"
})

adult_young_df = adult_young_df[(adult_young_df['year'] >= 1990) & (adult_young_df['year'] <= 2023)]



adult_young_df
```

```python
import pandas as pd 

illiterate_df = pd.read_csv("https://ourworldindata.org/grapher/literate-and-illiterate-world-population.csv?v=1&csvType=full&useColumnShortNames=true",storage_options = {"user-agent":"Our World In Data data fetch/1.0"})

illiterate_population_Data = pd.DataFrame(illiterate_df)

illiterate_population_Data['entity'] = illiterate_population_Data['entity'].str.strip()

illiterate_population_Data = illiterate_df.dropna(subset = ['code'])

illiterate_population_Data = illiterate_population_Data[(illiterate_population_Data['year'] >= 1990) & (illiterate_population_Data['year'] <= 2023)]



illiterate_population_Data
```

```python
illiterate_population_Data.isnull().sum()
```

```python
illiterate_population_Data.duplicated().sum()
```

```python
import pandas as pd

gdp_schooling_df = pd.merge(gdp_percapita,average_years_of_scholling, on = ["entity","year","code"])

gdp_schooling_df = gdp_schooling_df[(gdp_schooling_df['year'] >= 1990) & (gdp_schooling_df['year'] <= 2023)]

gdp_schooling_df = gdp_schooling_df.dropna(subset=['mf_youth_and_adults__15_64_years__average_years_of_education'])

gdp_schooling_df['entity'] = gdp_schooling_df['entity'].str.strip()

gdp_schooling_df = gdp_schooling_df.drop(columns=['literacy_rate'])

gdp_schooling_df = gdp_schooling_df.rename(columns = {
    "mf_youth_and_adults__15_64_years__average_years_of_education":"average_years_of_education",
     "ny_gdp_pcap_pp_kd":"gdp_percapita"})

gdp_schooling_df
```

```python
unusual_rows_ay = adult_young_df[
    (adult_young_df['adult_literacy_rate'] < 0) |
    (adult_young_df['adult_literacy_rate'] > 100) |
    (adult_young_df['youth_literacy_rate_male'] < 0) |
    (adult_young_df['youth_literacy_rate_male'] > 100) |
    (adult_young_df['youth_literacy_rate_female'] < 0) |
    (adult_young_df['youth_literacy_rate_female'] > 100)
]


unusual_rows_ay
```

```python
unusual_rows_gdp = gdp_schooling_df[
    (gdp_schooling_df['gdp_percapita'] < 0) |
    (gdp_schooling_df['average_years_of_education'] < 0) |
    (gdp_schooling_df['average_years_of_education'] > 25)
]

unusual_rows_gdp
```

```python
unusual_rows_il = illiterate_population_Data[
    (illiterate_population_Data['literacy_rate'] +
     illiterate_population_Data['illiteracy_rate'] < 99) |
    (illiterate_population_Data['literacy_rate'] +
     illiterate_population_Data['illiteracy_rate'] > 101)
]

unusual_rows_il
```
