```python

gdp_schooling_df = gdp_schooling_df.merge(
    illiterate_population_Data[['entity','year','literacy_pct']],
    on=['entity','year'],
    how='left'
)

max_years = gdp_schooling_df["average_years_of_education"].max()


gdp_schooling_df["gdp_per_schooling_year"] = (
    gdp_schooling_df["gdp_percapita"] /
    gdp_schooling_df["average_years_of_education"]
)

gdp_schooling_df["education_index"] = (
    (gdp_schooling_df["literacy_pct"] / 100) *
    (gdp_schooling_df["average_years_of_education"] / max_years)
)

gdp_schooling_df
```

```python
import pandas as pd

top5_literacy = """
SELECT entity,
       adult_literacy_rate
FROM literacy_rates
WHERE year = 2020
ORDER BY adult_literacy_rate DESC
LIMIT 5;
"""

cpa_cursor.execute(top5_literacy)

top5_literacy_result = cpa_cursor.fetchall()

df_top5_literacy = pd.DataFrame(
    top5_literacy_result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_top5_literacy #top 5 countries
```

```python
female_youth_literacy_lesser_than_80_percentage = """
SELECT entity,
       year,
       youth_literacy_rate_female
FROM literacy_rates
WHERE youth_literacy_rate_female < 80
ORDER BY youth_literacy_rate_female;
"""

cpa_cursor.execute(female_youth_literacy_lesser_than_80_percentage)

female_youth_literacy_result = cpa_cursor.fetchall()

df_female_literacy = pd.DataFrame(
    female_youth_literacy_result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_female_literacy #youth female literaccy rate lesser than 80%
```

```python
avg_region_literacy = """
SELECT owid_region,
       ROUND(AVG(adult_literacy_rate), 2) AS avg_adult_literacy
FROM literacy_rates
GROUP BY owid_region
ORDER BY avg_adult_literacy DESC;
"""

cpa_cursor.execute(avg_region_literacy)

avg_region_result = cpa_cursor.fetchall()

df_avg_region = pd.DataFrame(
    avg_region_result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_avg_region #average literacy adult rate per region
```

```python
import pandas as pd

top10_illiteracy = """
SELECT entity,
       illiteracy_pct
FROM illiteracy_population
WHERE year = (
    SELECT MAX(year)
    FROM illiteracy_population
)
ORDER BY illiteracy_pct DESC
LIMIT 10;
"""

cpa_cursor.execute(top10_illiteracy)

result = cpa_cursor.fetchall()

df_top10_illiteracy = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_top10_illiteracy #Top 10 countries with largest illiterate population in the last year
```

```python
import pandas as pd

schooling_gdp = """
SELECT entity,
       year,
       average_years_of_education,
       gdp_percapita
FROM gdp_schooling
WHERE average_years_of_education > 7
AND gdp_percapita < 5000
ORDER BY average_years_of_education DESC;
"""

cpa_cursor.execute(schooling_gdp)

result = cpa_cursor.fetchall()

df_schooling_gdp = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_schooling_gdp # Find countries with avg_years_schooling > 7 and gdp_per_capita < 5000.
```

```python
import pandas as pd

gdp_schooling_rank = """
SELECT entity,
       gdp_per_schooling_year,
       RANK() OVER(
           ORDER BY gdp_per_schooling_year DESC
       ) AS ranking
FROM gdp_schooling
WHERE year = 2020;
"""

cpa_cursor.execute(gdp_schooling_rank)

result = cpa_cursor.fetchall()

df_gdp_schooling_rank = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_gdp_schooling_rank  # Rank countries by GDP per schooling for the year 2020
```

```python
import pandas as pd

global_schooling_avg = """
SELECT year,
       ROUND(AVG(average_years_of_education),2)
       AS avg_schooling_years
FROM gdp_schooling
GROUP BY year
ORDER BY year;
"""

cpa_cursor.execute(global_schooling_avg)

result = cpa_cursor.fetchall()

df_global_schooling_avg = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_global_schooling_avg #Find global average schooling years per year.
```

```python
import pandas as pd

high_gdp_but_low_schooling_year = """
SELECT entity,
       gdp_percapita,
       average_years_of_education
FROM gdp_schooling
WHERE year = 2020
AND average_years_of_education < 6
ORDER BY gdp_percapita DESC
LIMIT 10;
"""

cpa_cursor.execute(high_gdp_but_low_schooling_year)

result = cpa_cursor.fetchall()

df_high_gdp_but_low_schooling_year = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_high_gdp_but_low_schooling_year #List top 10 countries in 2020 with highest GDP per capita but lowest average years of schooling(less than 6).
```

```python
import pandas as pd

high_illiterate_population_but_10years_of_schooling = """
SELECT g.entity,
       g.year,
       g.average_years_of_education,
       i.illiteracy_pct
FROM gdp_schooling g
JOIN illiteracy_population i
ON g.entity = i.entity
AND g.year = i.year
WHERE g.average_years_of_education > 10
AND i.illiteracy_pct > 20
ORDER BY i.illiteracy_pct DESC;
"""

cpa_cursor.execute(high_illiterate_population_but_10years_of_schooling)

result = cpa_cursor.fetchall()

high_illiterate_population_but_10years_of_schooling = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

high_illiterate_population_but_10years_of_schooling #Show countries where the illiterate population is high despite having more than 10 average years of schooling.
```

```python
import pandas as pd

india_literacyrates_vs_gdp_per_capita = """
SELECT year,
       literacy_pct,
       gdp_percapita
FROM gdp_schooling
WHERE entity = 'India'
AND year BETWEEN 2000 AND 2020
ORDER BY year;
"""

cpa_cursor.execute(india_literacyrates_vs_gdp_per_capita)

result = cpa_cursor.fetchall()

df_india_literacyrates_vs_gdp_per_capita = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_india_literacyrates_vs_gdp_per_capita #Compare literacy rates and GDP per capita growth for a selected country over the last 20 years. (country of your choice)
```

```python
import pandas as pd

literacyrate_male_vs_female = """
SELECT l.entity,
       g.gdp_percapita,
       l.youth_literacy_rate_male,
       l.youth_literacy_rate_female,
       ABS(
           l.youth_literacy_rate_male -
           l.youth_literacy_rate_female
       ) AS literacy_gap
FROM literacy_rates l
JOIN gdp_schooling g
ON l.entity = g.entity
AND l.year = g.year
WHERE g.year = 2020
AND g.gdp_percapita > 30000
ORDER BY literacy_gap DESC;
"""

cpa_cursor.execute(literacyrate_male_vs_female)

result = cpa_cursor.fetchall()

df_literacyrate_male_vs_female = pd.DataFrame(
    result,
    columns=[col[0] for col in cpa_cursor.description]
)

df_literacyrate_male_vs_female #Show the difference between youth literacy male and female rates for countries with GDP per capita above $30,000 in 2020.
```
