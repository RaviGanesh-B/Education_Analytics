```python
import pandas as pd
import requests

adult_young_df = pd.merge(adult_literacy_rate_df,young_literacy_rate_df, on = ["entity","year","code"])

adult_young_df = adult_young_df.dropna(subset=['code'])

adult_young_df['entity'] = adult_young_df['entity'].str.strip()

adult_young_df["owid_region"] = adult_young_df["owid_region"].fillna("unknown")

adult_young_df = adult_young_df.rename(columns = {
    "adult_literacy_rate__population_15plus_years__both_sexes__pct__lr_ag15t99":"adult_literacy_rate",
    "youth_literacy_rate__population_15_24_years__male__pct__lr_ag15t24_m":"youth_literacy_rate_male",
    "youth_literacy_rate__population_15_24_years__female__pct__lr_ag15t24_f":"youth_literacy_rate_female"
})

literacy_df = adult_young_df[(adult_young_df['year'] >= 1990) & (adult_young_df['year'] <= 2023)]

literacy_df
```

```python

gdp_schooling_df = gdp_schooling_df.merge(
    illiterate_population_Data[['entity','year','literacy_pct']],
    on=['entity','year'],
    how='left'
)

max_years = gdp_schooling_df["average_years_of_education"].max()


gdp_schooling_df["gdp_per_schooling_year"] = (
    gdp_schooling_df["gdp_percapita"] /
    gdp_schooling_df["average_years_of_education"]
)

gdp_schooling_df["education_index"] = (
    (gdp_schooling_df["literacy_pct"] / 100) *
    (gdp_schooling_df["average_years_of_education"] / max_years)
)

gdp_schooling_df
```

```python
(gdp_schooling_df["owid_region_x"] == 
 gdp_schooling_df["owid_region_y"]).all()
```

```python
gdp_schooling_df = gdp_schooling_df.drop(columns = {"owid_region_y"})

gdp_schooling_df = gdp_schooling_df.rename(columns = {"owid_region_x":"owid_region"})
```

```python
gdp_schooling_df["literacy_pct"] = gdp_schooling_df.groupby("entity")["literacy_pct"].transform(lambda x : x.fillna(x.mean()))
```

```python
gdp_schooling_df["education_index"] = gdp_schooling_df.groupby("entity")["education_index"].transform(lambda x : x.fillna(x.mean()))

gdp_schooling_df
```

```python
gdp_schooling_df = gdp_schooling_df.sort_values(["entity", "year"])

gdp_schooling_df["literacy_pct"] = (
    gdp_schooling_df.groupby("entity")["literacy_pct"]
    .transform(lambda x: x.ffill().bfill())
)
```
